```python
import numpy as np

def estimate_youngs_modulus(L, r, f, rho):
    """
    Estimate Young's modulus for a cantilever beam
    given length, radius, frequency, and density.

    Parameters:
    -----------
    L   : float
        Length of the beam (m)
    r   : float
        Radius of circular cross-section (m)
    f   : float
        Measured first natural frequency (Hz)
    rho : float
        Density (kg/m^3)

    Returns:
    --------
    E : float
        Estimated Young's modulus (Pa)
    """

    # First bending mode constant for a cantilever
    beta1 = 1.8751

    # Cross-sectional area and second moment of area
    A = np.pi * r**2
    I = np.pi * r**4 / 4

    # Formula for Young's modulus
    E = ((2 * np.pi * f)**2 * rho * A * L**4) / (beta1**4 * I)
    return E

```

```python

# Example: your provided parameters
L = 0.292      # m
r = 0.003      # m
f = 1.38       # Hz
rho = 1100     # kg/m^3

E = estimate_youngs_modulus(L, r, f, rho)

print(f"Estimated Young's modulus: {E:.2e} Pa ({E/1e6:.2f} MPa)")
```
