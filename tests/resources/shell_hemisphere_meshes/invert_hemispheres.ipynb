```python
import os


def invert_z(in_path: str, out_path: str) -> None:
    """Read a dismech input .txt file and write a copy with z -> -z for all nodes.

    Mirrors the parsing convention in Geometry.from_txt: lines starting with '#'
    are comments, lines starting with '*' are headers (case-insensitive),
    other lines are comma-separated values.
    """
    valid_headers = {"*nodes", "*edges", "*triangles"}
    cur_h = None

    with open(in_path, "r") as f_in, open(out_path, "w") as f_out:
        for raw in f_in:
            line = raw.strip()

            if not line or line.startswith("#"):
                f_out.write(raw)
                continue

            if line.startswith("*"):
                if line.lower() not in valid_headers:
                    raise ValueError(f"Unknown header: {line}")
                cur_h = line.lower()
                f_out.write(raw)
                continue

            if cur_h == "*nodes":
                vals = [v.strip() for v in line.split(",")]
                if len(vals) != 3:
                    raise ValueError(f"{vals} should have 3 values")
                x, y, z = (float(v) for v in vals)
                f_out.write(f"{x},{y},{-z}\n")
            else:
                f_out.write(raw)


inputs = [
    # "hemisphere_n50_r0.1_mh0.001_nh0.txt",
    # "hemisphere_n50_r0.1_mh0.001_nh1.txt",
    "hemisphere_n146_r0.1_mh0.001_nh0.txt",
]

here = os.path.dirname(os.path.abspath("__file__"))
for name in inputs:
    in_path = os.path.join(here, name)
    base, ext = os.path.splitext(name)
    out_path = os.path.join(here, f"{base}_inverted{ext}")
    invert_z(in_path, out_path)
    print(f"wrote {out_path}")
```
